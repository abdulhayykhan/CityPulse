# 01 · Audio Data Check
Sanity-check the UrbanSound8K and ESC-50 datasets before feature extraction.

```python
# Cell 1 – interpreter & imports
import sys
print(sys.executable)

import librosa
import sklearn
import soundfile
print('imports OK')
```

```python
# Cell 2 – dataset paths
from pathlib import Path

ROOT  = Path.cwd().parent
US8K  = ROOT / 'data' / 'raw' / 'UrbanSound8K'
ESC50 = ROOT / 'data' / 'raw' / 'ESC-50'

print('ROOT :', ROOT)
print('US8K exists :', US8K.exists())
print('ESC50 exists:', ESC50.exists())
```

```python
# Cell 3 – metadata overview
import pandas as pd

# UrbanSound8K
us8k_meta = pd.read_csv(US8K / 'metadata' / 'UrbanSound8K.csv')
print('=== UrbanSound8K ===')
print('Shape:', us8k_meta.shape)
print('\nClass value_counts:')
print(us8k_meta['class'].value_counts())
print('\nFold value_counts (sorted by fold):')
fold_counts = us8k_meta['fold'].value_counts().sort_index()
print(fold_counts)
print('Fold total:', fold_counts.sum())

# ESC-50
esc_meta = pd.read_csv(ESC50 / 'meta' / 'esc50.csv')
print('\n=== ESC-50 ===')
print('Shape:', esc_meta.shape)
print('Unique categories:', esc_meta['category'].nunique())
```

```python
# Cell 4 – fold1 contents
fold1_path = US8K / 'audio' / 'fold1'
fold1_items = list(fold1_path.iterdir())

total_entries = len(fold1_items)
wav_files     = [f for f in fold1_items if f.suffix.lower() == '.wav']
non_wav       = [f.name for f in fold1_items if f.suffix.lower() != '.wav']

print(f'fold1 total entries : {total_entries}')
print(f'fold1 .wav files    : {len(wav_files)}')
print(f'fold1 non-.wav      : {non_wav}')
```

```python
# Cell 5 – load audio clips
import librosa
import numpy as np

TARGET = '100032-3-0-0.wav'

# look up fold from CSV (never hardcoded)
row  = us8k_meta.loc[us8k_meta['slice_file_name'] == TARGET].iloc[0]
fold = row['fold']
us8k_clip_path = US8K / 'audio' / f'fold{fold}' / TARGET

y_us, sr_us = librosa.load(us8k_clip_path, sr=22050, mono=True)
print('=== UrbanSound8K clip ===')
print('Path  :', us8k_clip_path)
print('Shape :', y_us.shape)
print('SR    :', sr_us)
print('Duration (s):', round(len(y_us) / sr_us, 4))

# ESC-50 clip
esc_clip_path = ESC50 / 'audio' / '1-100032-A-0.wav'
y_esc, sr_esc = librosa.load(esc_clip_path, sr=22050, mono=True)
print('\n=== ESC-50 clip ===')
print('Path  :', esc_clip_path)
print('Shape :', y_esc.shape)
print('SR    :', sr_esc)
print('Duration (s):', round(len(y_esc) / sr_esc, 4))
```
